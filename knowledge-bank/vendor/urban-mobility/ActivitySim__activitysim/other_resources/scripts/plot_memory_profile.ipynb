```python
import pandas as pd
import plotly.express as px
from datetime import datetime
```

```python
df = pd.read_csv(r"C:\path\to\memory_profile.csv")
#convert full_rss bytes to gb
df['full_rss_gb'] = df['full_rss'] / 1000000000

#process time column to only get day and hour and minute and second
df['time'] = pd.to_datetime(df['time'], format='%Y-%m-%d %H:%M:%S')

summary = df.groupby(['time','event']).full_rss_gb.max().reset_index()

#plot full_rss over time usign line plot and segmented by event
fig = px.line(summary, x='time', y='full_rss_gb', color='event')
fig.update_layout(title='Memory Profile', xaxis_title='Time', yaxis_title='Max. Memory in GB (full rss)')
#add date to the html file name below
current_date = datetime.now().strftime('%Y%m%d')
fig.write_html(f"memory_profile{current_date}.html")
```

```python
fig.show()
```
